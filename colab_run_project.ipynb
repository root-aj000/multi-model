This notebook installs dependencies, updates the dataset path in the config, and runs the training script from the `multi-model` project.

```python
# Install required dependencies from the project requirements file
import os
from pathlib import Path

repo_root = Path('/content/backup')
requirements_path = repo_root / 'multi-model' / 'requirements.txt'
print('Requirements path:', requirements_path)

if requirements_path.exists():
    !python -m pip install -r "{requirements_path}"
else:
    raise FileNotFoundError(f'Requirements file not found: {requirements_path}')
```

## Purpose of this notebook
This notebook performs the following tasks:

1. Installs Python dependencies required by the `multi-model` project.
2. Updates `DATASET_ROOT` in `multi-model/configs/model/model_config.json` to point to the dataset path available in Colab.
3. Runs the training script using the updated config.
4. Stores trained models and logs under the notebook workspace.

```python
import json
from pathlib import Path

config_file = Path('/content/backup/multi-model/configs/model/model_config.json')
print('Config file:', config_file)

if not config_file.exists():
    raise FileNotFoundError(f'Config file not found: {config_file}')

config = json.loads(config_file.read_text())
config['DATASET_ROOT'] = '/kaggle/input/datasets/anishjagdaleaj/testdataset'

config_file.write_text(json.dumps(config, indent=2))
print('Updated DATASET_ROOT to:', config['DATASET_ROOT'])
```
